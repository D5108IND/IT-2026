```python
def evaluate_effectiveness(losses, enemy_strength):
    if enemy_strength == 0:
        return 0

    return losses / enemy_strength


print(evaluate_effectiveness(66, 10))
```

```python
def average_losses(*days):
    if not days:
        return 0

    return sum(days) / len(days)


print(average_losses(99, 35, 32, 66, 89))
```
