```python
import csv

data = [
    ["Південь", 78],
    ["Північ", 34],
    ["Захід", 45],
    ["Схід", 189],
]

with open("losses.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerows(data)

print("Файл losses.csv створено!")
```

```python
data = [
    {"підрозділ": "А", "втрати": 10},
    {"підрозділ": "А", "втрати": 20},
    {"підрозділ": "Б", "втрати": 15},
    {"підрозділ": "Б", "втрати": 25},
    {"підрозділ": "В", "втрати": 30}
]

середні = {}

for запис in data:
    підрозділ = запис["підрозділ"]
    втрати = запис["втрати"]

    if підрозділ not in середні:
        середні[підрозділ] = []

    середні[підрозділ].append(втрати)

for підрозділ, значення in середні.items():
    середнє = sum(значення) / len(значення)
    print(f"{підрозділ}: {середнє:.2f}")
```

```python
import csv

data = [
    {"підрозділ": "А", "втрати": 10, "рівень загрози": "Низький"},
    {"підрозділ": "А", "втрати": 20, "рівень загрози": "Середній"},
    {"підрозділ": "Б", "втрати": 15, "рівень загрози": "Високий"},
    {"підрозділ": "Б", "втрати": 25, "рівень загрози": "Високий"},
]

результат = {}

for запис in data:
    підрозділ = запис["підрозділ"]

    if підрозділ not in результат:
        результат[підрозділ] = {
            "втрати": [],
            "рівень загрози": запис["рівень загрози"]
        }

    результат[підрозділ]["втрати"].append(запис["втрати"])

with open("аналіз_втрат.csv", "w", newline="", encoding="utf-8-sig") as файл:
    writer = csv.writer(файл)

    writer.writerow(["Підрозділ", "Середні втрати", "Рівень загрози"])

    for підрозділ, дані in результат.items():
        середні_втрати = sum(дані["втрати"]) / len(дані["втрати"])

        writer.writerow([
            підрозділ,
            round(середні_втрати, 2),
            дані["рівень загрози"]
        ])

print("Результат збережено у файл аналіз_втрат.csv")
```
