```python
content = "report.txt"

with open ("some.txt", "w") as f:
    f.write(content)
```

```python
with open ("report.txt","a") as f:
    f.write(content)
```

```python
with open("report.txt", "r", encoding="utf-8") as file:
    text = file.read()

for line in text.splitlines():
    if "Втрати противника:" in line:
        losses = line.split(":")[1].strip()
        print("Кількість втрат противника:", losses)
```

```python
report = {
    "Втрати противника": 133,
    "Знищено техніки": 44,
    "Захоплено позицій": 4
}

with open("summary.txt", "w", encoding="utf-8") as file:
    file.write("Підсумок за день\n")
    file.write("================\n")
    
    for key, value in report.items():
        file.write(f"{key}: {value}\n")
```

```python
from datetime import datetime

def write_log(message):
    with open("log.txt", "a", encoding="utf-8") as log:
        time = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        log.write(f"[{time}] {message}\n")


with open("report.txt", "r", encoding="utf-8") as file:
    text = file.read()

write_log("Зчитано файл report.txt")


with open("losses.csv", "r", encoding="utf-8") as file:
    data = file.read()

write_log("Зчитано файл losses.csv")

print("Дані зчитано. Інформацію записано до log.txt.")
```
