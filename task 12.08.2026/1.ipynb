```python
1. Вивести повідомлення про рівень втрат противника на основі введеного числа.
2. Порахувати середній рівень готовності трьох підрозділів.
3. Створити цикл, що виводить день і номер сектора з номером розрахунку.
```

```python
vtraty = int(input("Введіть кількість втрат: "))

if vtraty < 20:
    print("Рівень втрат: низький")
elif vtraty <= 50:
    print("Рівень втрат: середній")
else:
    print("Рівень втрат: високий")
```

```python
gotovnist1 = 80
gotovnist2 = 65
gotovnist3 = 90

serednya = (gotovnist1 + gotovnist2 + gotovnist3) / 3

print("Середній рівень готовності:", serednya, "%")
```

```python
for den in range(1, 6):
    for sektor in range(1, 4):
        for rozrakhunok in range(1, 3):
            print("День:", den,
                  "Сектор:", sektor,
                  "Номер розрахунку:", rozrakhunok)
```
