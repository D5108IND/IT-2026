```python
losses = int(input("Введіть кількість втрат:65"))

if losses == 0:
    print("Втрати противника відсутні")
elif losses <= 50:
    print("низікі")
elif losses <= 100:
    print("високі")
else:
    print("Рівень втрат критичний")
```
