```python
mark = 65

if mark >= 90:
    print("Excellent")
    print("You have an A grade")
elif mark >= 80:
    print("Wery good")
    print("You have a B grade")
elif mark >= 70:
    print("Good")
    print("You have a C grade")
else:
    print("Not bad")
    print("You have a D grade.")
```
